```python
import sys
sys.path.append("..")
import pandas as pd
from src.pipeline_utils import get_engine
from analysis.volatility_renewable_share import load_prices
from src.entsoe_client import PRICE_ZONES, FLOW_PAIRS
```

```python

def load_prices(engine, zones):
    """Load day-ahead prices for the given zones and parse timestamps."""
    prices = pd.read_sql("SELECT * FROM prices", engine)
    prices["timestamp"] = pd.to_datetime(prices["timestamp"])
    return prices[prices["zone"].isin(zones)]
```

```python


def load_flows(engine, zones):
    """Load cross-border flows for the given zones and parse timestamps."""
    flows = pd.read_sql("SELECT * FROM cross_border_flows", engine)
    flows["timestamp"] = pd.to_datetime(flows["timestamp"])
    return flows[flows["zone"].isin(zones)]
```
